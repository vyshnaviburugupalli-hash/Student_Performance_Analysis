```python
#using pandas and numpy

import pandas as pd
import numpy as np

#loading the dataset

df = pd.read_csv("Student_data.csv")

print("STUDENT PERFORMANCE ANALYSIS")
print("=" * 50)

```

```python
#Displaying the data

print("\n1. COMPLETE STUDENT DATA")
display(df)

print("\n2. FIRST FIVE RECORDS")
display(df.head())

print("\n3. COLUMN NAMES")
print(df.columns.tolist())

print("\n4. NUMBER OF ROWS AND COLUMNS")
print(df.shape)
```

```python
student_count = len(df)

student_names = df["Name"].tolist()
student_ids = df["Student_Id"].tolist()

subjects = [
    "SQL",
    "Python",
    "Artificial Intelligence",
    "Big Data Analytics",
    "Data Structures"
]

# Convert subject marks and attendance into numeric data
for column in subjects + ["Attendance"]:
    df[column] = pd.to_numeric(df[column], errors="coerce")

if df.empty:
    raise ValueError("The CSV file contains no student records.")

if df[["Student_Id", "Name"] + subjects + ["Attendance"]].isnull().any().any():
    raise ValueError("Please correct missing or non-numeric required values.")

if df["Student_Id"].duplicated().any():
    raise ValueError("Duplicate student IDs found.")

if not df[subjects].apply(
    lambda column: column.between(0, 100).all()
).all():
    raise ValueError("Subject marks must be between 0 and 100.")

if not df["Attendance"].between(0, 100).all():
    raise ValueError("Attendance must be between 0 and 100.")

print("\nTotal students:", student_count)
print("Student names:", student_names)
print("Student IDs:", student_ids)
```

```python
df["Total_Marks"] = df.apply(calculate_total, axis=1)

df["Average_Marks"] = df.apply(calculate_average, axis=1)

df["Grade"] = df["Average_Marks"].apply(assign_grade)

# Use a loop to calculate each student's result
results = []

for index, row in df.iterrows():
    result = check_result(row)
    results.append(result)

df["Result"] = results

print("\nPROCESSED STUDENT RESULTS")
print("=" * 50)

display(
    df[[
        "Student_Id",
        "Name",
        "Total_Marks",
        "Average_Marks",
        "Grade",
        "Result",
        "Attendance"
    ]].round(2)
)
```

```python
# Convert average marks into a NumPy array
averages = df["Average_Marks"].to_numpy()

overall_average = np.mean(averages)
highest_average = np.max(averages)
lowest_average = np.min(averages)
median_average = np.median(averages)
standard_deviation = np.std(averages)

highest_student = df.loc[df["Average_Marks"].idxmax()]
lowest_student = df.loc[df["Average_Marks"].idxmin()]

print("\nOVERALL CLASS PERFORMANCE")
print("=" * 50)

print("Total students:", student_count)
print(f"Overall class average: {overall_average:.2f}")
print(f"Highest student average: {highest_average:.2f}")
print(f"Lowest student average: {lowest_average:.2f}")
print(f"Median student average: {median_average:.2f}")
print(f"Standard deviation: {standard_deviation:.2f}")

print(
    f"Highest-performing student: {highest_student['Name']} "
    f"({highest_average:.2f})"
)

print(
    f"Lowest-performing student: {lowest_student['Name']} "
    f"({lowest_average:.2f})"
)
```

```python
#pass or fail analysis

pass_count = int((df["Result"] == "Pass").sum())
fail_count = int((df["Result"] == "Fail").sum())

pass_rate = np.divide(pass_count, student_count) * 100
fail_rate = np.divide(fail_count, student_count) * 100

print("\nPASS AND FAIL ANALYSIS")
print("=" * 50)

print("Students passed:", pass_count)
print("Students failed:", fail_count)
print(f"Pass percentage: {pass_rate:.2f}%")
print(f"Fail percentage: {fail_rate:.2f}%")

print("\nPassed students:")
display(df.loc[df["Result"] == "Pass", ["Student_Id", "Name", "Average_Marks"]])

print("\nFailed students:")
display(df.loc[df["Result"] == "Fail", ["Student_Id", "Name", "Average_Marks"]])
```

```python
#subject wise average

subject_averages = df[subjects].mean()

print("\nSUBJECT-WISE AVERAGE MARKS")
print("=" * 50)

display(
    subject_averages.rename("Average_Marks").to_frame().round(2)
)

best_subject = subject_averages.idxmax()
lowest_subject = subject_averages.idxmin()

print("Best-performing subject:", best_subject)
print("Lowest-performing subject:", lowest_subject)
```

```python
# Use a loop to analyse every subject
for subject in subjects:
    subject_marks = df[subject].to_numpy()

    print(f"\nSubject: {subject}")
    print(f"Average marks: {np.mean(subject_marks):.2f}")
    print(f"Highest marks: {np.max(subject_marks)}")
    print(f"Lowest marks: {np.min(subject_marks)}")

    highest_index = df[subject].idxmax()
    highest_name = df.loc[highest_index, "Name"]

    print(
        f"Top student: {highest_name} "
        f"({df.loc[highest_index, subject]} marks)"
    )
```

```python
#grade distribution
grade_distribution = df["Grade"].value_counts().reindex(
    ["A", "B", "C", "D", "F"],
    fill_value=0
)

print("\nGRADE DISTRIBUTION")
print("=" * 50)

display(grade_distribution.rename("Student_Count").to_frame())

#attendance analysis
average_attendance = np.mean(df["Attendance"])

print("\nATTENDANCE ANALYSIS")
print("=" * 50)

print(f"Average attendance: {average_attendance:.2f}%")

# Identify students with attendance below 75%
low_attendance = df[df["Attendance"] < 75]

print("\nStudents with attendance below 75%:")

if low_attendance.empty:
    print("No students have attendance below 75%.")
else:
    display(
        low_attendance[["Student_Id", "Name", "Attendance"]]
    )


#final report

print("\n" + "=" * 50)
print("FINAL STUDENT PERFORMANCE REPORT")
print("=" * 50)

print(f"Total students analysed: {student_count}")
print(f"Overall class average: {overall_average:.2f}")
print(f"Highest average: {highest_average:.2f}")
print(f"Lowest average: {lowest_average:.2f}")
print(f"Best subject: {best_subject}")
print(f"Lowest-performing subject: {lowest_subject}")
print(f"Students passed: {pass_count}")
print(f"Students failed: {fail_count}")
print(f"Pass rate: {pass_rate:.2f}%")
print(f"Average attendance: {average_attendance:.2f}%")

print("\nTop-performing students:")

# Fix: Define top_students by sorting the dataframe by Average_Marks in descending order
top_students = df.sort_values(by="Average_Marks", ascending=False).head(5)

for rank, (_, row) in enumerate(top_students.iterrows(), start=1):
    print(
        f"{rank}. {row["Name"]} - "
        f"Average: {row["Average_Marks"]:.2f}, "
        f"Grade: {row["Grade"]}, "
        f"Result: {row["Result"]}"
    )
```

```python
#export final results


df.to_csv("Student_Performance_Results.csv", index=False)

subject_averages.rename("Average_Marks").to_csv(
    "Subject_Performance_Analysis.csv"
)

grade_distribution.rename("Student_Count").to_csv(
    "Grade_Distribution.csv"
)

print("\nAnalysis completed successfully.")
print("Output files created:")
print("1. Student_Performance_Results.csv")
print("2. Subject_Performance_Analysis.csv")
print("3. Grade_Distribution.csv")
```
